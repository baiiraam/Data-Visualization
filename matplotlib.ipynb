```python
# Remove the symbols, the final two numbers from the 'Price' column and convert it to numbers
car_sales["Price"] = car_sales["Price"].apply(lambda x: x[1:].replace(',', '').replace('.', '')).astype('int')
# Add a column called 'Total Sales' to car_sales which cumulatively adds the 'Price' column
car_sales["Total Sales"] = car_sales["Price"].cumsum()

# Add a column called 'Sale Date' which lists a series of successive dates starting from today (your today)
car_sales["Sale Date"] = pd.date_range('today', periods=len(car_sales), freq='D').normalize()
# View the car_sales DataFrame
car_sales.head()
```

```python
fig, ax = plt.subplots(nrows=3, ncols=1, figsize=(12, 14))

fig.tight_layout(pad=6.0)

ax[0].scatter(df.groupby("iyear").count()["success"].reset_index().iyear.values,\
         df.groupby("iyear").count()["success"].reset_index().success.values);

ax[0].set_xlabel('year', fontsize='large')
ax[0].set_ylabel('terror attempts', fontsize='large')
ax[0].set_xlim(1967, 2020)
ax[0].set_title('Terror attempts throughout the years', fontsize='xx-large')


ax[1].bar(["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],\
        df.groupby('imonth').count()["success"].reset_index().iloc[1:, :]["success"].values);
ax[1].set_xlabel('Month', fontsize='large')
ax[1].set_ylabel('Terror Attempts', fontsize='large')
ax[1].set_title('Terror attempts through months', fontsize='xx-large')


ax[2].bar(df.groupby('iday').count()["success"].reset_index().iloc[1:, :].iday.values,\
df.groupby('iday').count()["success"].reset_index().iloc[1:, :].success.values);
ax[2].set_xlabel('day of the month', fontsize='large')
ax[2].set_ylabel('terror attempts', fontsize='large')
ax[2].set_title('terror attempts occurrency through days', fontsize='xx-large')


for i in range(3):
  ax[i].yaxis.set_major_formatter(mpl.ticker.StrMethodFormatter('{x:,.0f}'))
```
